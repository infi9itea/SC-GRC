This is the only cache benchmark for the hybrid. It uses **exactly 40 queries total** from the same 333-item evaluation set. It is intended to fit in one Kaggle saved-version run. It does not run the old 200-query cache benchmark.

```python
# ========================= USER CONFIG =========================
MODEL_ID_SLM = "Qwen/Qwen2.5-3B-Instruct"
SLM_ADAPTER_DIR = "/kaggle/input/datasets/mrnotalent/ewu-qwen-adapter-checkpoint1632"  # EDIT if needed
KB_DIR = "/kaggle/input/datasets/mohuaakter/ewu-dataset-jsonl-pairs/KB/KB"  # KB ONLY; no scraped data
GROUND_TRUTH_XLSX = "/kaggle/input/datasets/mohuaakter/ewu-dataset-jsonl-pairs/University_Chatbot_Questions_1110_GROUND_TRUTHS_VERIFIED.xlsx"  # EDIT
INPUT_STATE_DIR = "/kaggle/input/datasets/bxgdhdgsg/555555/sc_grc_state"  # EDIT: chains from the completed 333-row router state (same chain used by Notebook 08), NOT directly from Notebook 04
WORK_STATE_DIR = "/kaggle/working/sc_grc_state"

SEED = 42
TOP_K_DENSE, TOP_K_SPARSE, TOP_K_FINAL = 10, 10, 4
K_SAMPLES = 3
SC_GRC_TEMPERATURE = 0.7
SC_GRC_TOP_P = 0.9
QUALITY_FLOOR_BERTSCORE = 0.85
MAX_NEW_TOKENS = 200

MODEL_ID_LLM = "Qwen/Qwen2.5-7B-Instruct"
LLM_ADAPTER_DIR = "/kaggle/input/datasets/mrnotalent/7bbbbb/ewu7b/adapter"  # EDIT
```

```python

# Exactly 40 queries. No 200-query stream.
import os, json, re, time, shutil
from collections import Counter
import numpy as np, pandas as pd, torch, faiss, joblib
from tqdm.auto import tqdm
from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig
from peft import PeftModel
from rank_bm25 import BM25Okapi
from sentence_transformers import SentenceTransformer
from FlagEmbedding import FlagReranker

os.makedirs(WORK_STATE_DIR, exist_ok=True)
if INPUT_STATE_DIR and os.path.isdir(INPUT_STATE_DIR) and INPUT_STATE_DIR != WORK_STATE_DIR:
    for name in os.listdir(INPUT_STATE_DIR):
        src, dst = os.path.join(INPUT_STATE_DIR, name), os.path.join(WORK_STATE_DIR, name)
        if os.path.isfile(src) and not os.path.exists(dst): shutil.copy2(src, dst)

eval_df = pd.read_csv(os.path.join(WORK_STATE_DIR, 'eval_333.csv'))
sig = pd.read_csv(os.path.join(WORK_STATE_DIR, 'sc_grc_signals_with_router.csv')).sort_values('idx').reset_index(drop=True)
if len(sig) != 333: raise RuntimeError('Need completed 333-row router state (same chain used by Notebook 08)')
clf = joblib.load(os.path.join(WORK_STATE_DIR, 'sc_grc_router.joblib'))

kb_chunks = json.load(open(os.path.join(WORK_STATE_DIR, 'kb.json'), encoding='utf-8'))
kb_sources = json.load(open(os.path.join(WORK_STATE_DIR, 'kb_meta.json'), encoding='utf-8'))['sources']
dense_index = faiss.read_index(os.path.join(WORK_STATE_DIR, 'dense_index.faiss'))
bm25 = BM25Okapi([re.findall(r"[\w\u0980-\u09FF]+", c.lower()) for c in kb_chunks])
DEVICE = 'cuda' if torch.cuda.is_available() else 'cpu'
embedder = SentenceTransformer('BAAI/bge-m3', device=DEVICE); embedder.max_seq_length = 512
reranker = FlagReranker('BAAI/bge-reranker-v2-m3', use_fp16=True, device=DEVICE)

def embed_texts(texts):
    return np.asarray(embedder.encode(texts, batch_size=16, normalize_embeddings=False, show_progress_bar=False), dtype='float32')

def hybrid_retrieve(query):
    q = embed_texts([query])
    faiss.normalize_L2(q)

    _, di = dense_index.search(q, min(TOP_K_DENSE, len(kb_chunks)))

    ss = bm25.get_scores(re.findall(r"[\w\u0980-\u09FF]+", query.lower()))
    si = np.argsort(ss)[::-1][:min(TOP_K_SPARSE, len(kb_chunks))]

    ids = sorted(set(di[0].tolist()) | set(si.tolist()))
    pairs = [[query, kb_chunks[i]] for i in ids]

    raw_scores = reranker.compute_score(pairs, normalize=True)

    # Flatten to a 1-D scalar array (same fix as Notebook 04/08 -- raw output
    # shape/type from FlagReranker.compute_score varies with candidate count).
    rs = np.asarray(raw_scores, dtype=np.float32).reshape(-1)

    if len(rs) != len(ids):
        raise RuntimeError(
            f"Reranker returned {len(rs)} scores for {len(ids)} candidate chunks. "
            f"Raw shape={np.asarray(raw_scores).shape}"
        )

    order = np.argsort(rs)[::-1][:TOP_K_FINAL]

    return [
        {
            'source': kb_sources[ids[int(j)]],
            'chunk': kb_chunks[ids[int(j)]],
            'score': float(rs[int(j)])
        }
        for j in order
    ]

SYSTEM_PROMPT = ("You are the East West University (EWU) student support assistant. Answer the student's question directly, accurately, and helpfully. Do not invent university policies or facts. If the provided context is insufficient, say so clearly.")

def build_rag_prompt(query, retrieved):
    ctx = '\n\n'.join(f"[{r['source']}]\n{r['chunk']}" for r in retrieved) if retrieved else '(no relevant context retrieved)'
    user = f'Context:\n{ctx}\n\nQuestion: {query}\n\nAnswer using only the context above, following the system rules.'
    return f'<|im_start|>system\n{SYSTEM_PROMPT}<|im_end|>\n<|im_start|>user\n{user}<|im_end|>\n<|im_start|>assistant\n'

# Load both tiers for genuine cold hybrid execution.
bnb = BitsAndBytesConfig(load_in_4bit=True, bnb_4bit_use_double_quant=True, bnb_4bit_quant_type='nf4', bnb_4bit_compute_dtype=torch.float16)

slm_tok = AutoTokenizer.from_pretrained(SLM_ADAPTER_DIR, trust_remote_code=True, padding_side='right'); slm_tok.pad_token = slm_tok.pad_token or slm_tok.eos_token
slm_base = AutoModelForCausalLM.from_pretrained(MODEL_ID_SLM, quantization_config=bnb, device_map='auto', trust_remote_code=True, torch_dtype=torch.float16, attn_implementation='eager')
slm = PeftModel.from_pretrained(slm_base, SLM_ADAPTER_DIR).eval(); slm.config.use_cache = True

llm_tok = AutoTokenizer.from_pretrained(LLM_ADAPTER_DIR, trust_remote_code=True, padding_side='right'); llm_tok.pad_token = llm_tok.pad_token or llm_tok.eos_token
llm_base = AutoModelForCausalLM.from_pretrained(MODEL_ID_LLM, quantization_config=bnb, device_map='auto', trust_remote_code=True, torch_dtype=torch.float16, attn_implementation='sdpa')
llm = PeftModel.from_pretrained(llm_base, LLM_ADAPTER_DIR).eval(); llm.config.use_cache = True

@torch.no_grad()
def generate_slm(query, retrieved):
    p = build_rag_prompt(query, retrieved)
    e = slm_tok(p, return_tensors='pt', truncation=True, max_length=2048).to(slm.device)
    if torch.cuda.is_available(): torch.cuda.empty_cache()
    out = slm.generate(**e, max_new_tokens=MAX_NEW_TOKENS, do_sample=True, temperature=SC_GRC_TEMPERATURE,
                        top_p=SC_GRC_TOP_P, num_return_sequences=K_SAMPLES, pad_token_id=slm_tok.pad_token_id)
    n = e['input_ids'].shape[1]
    samples = [slm_tok.decode(o[n:], skip_special_tokens=True).strip() for o in out]
    del out
    if torch.cuda.is_available(): torch.cuda.empty_cache()
    return samples

@torch.no_grad()
def generate_llm(query, retrieved):
    p = build_rag_prompt(query, retrieved)

    # Truncate the RAG context, same VRAM-safety guard used in Notebook 08's
    # FT-LLM generation.
    e = llm_tok(p, return_tensors='pt', truncation=True, max_length=2048).to(llm.device)

    if torch.cuda.is_available(): torch.cuda.empty_cache()

    out = llm.generate(**e, max_new_tokens=MAX_NEW_TOKENS, do_sample=False, pad_token_id=llm_tok.pad_token_id)

    answer = llm_tok.decode(out[0][e['input_ids'].shape[1]:], skip_special_tokens=True).strip()

    del out
    if torch.cuda.is_available(): torch.cuda.empty_cache()

    return answer

def rep_and_consistency(samples):
    e = embed_texts(samples); faiss.normalize_L2(e)
    c = e.mean(axis=0, keepdims=True); faiss.normalize_L2(c)
    rep = samples[int(np.argmax((e @ c.T).reshape(-1)))]
    sims = [float(np.dot(e[i], e[j])) for i in range(len(samples)) for j in range(i + 1, len(samples))]
    return rep, float(np.mean(sims))

class SemanticCache:
    def __init__(self, thr=0.92): self.thr = thr; self.entries = []; self.exact = {}; self.index = None; self.stats = Counter()
    def rebuild(self):
        if not self.entries: self.index = None; return
        m = np.stack([e['emb'] for e in self.entries]).astype('float32'); faiss.normalize_L2(m)
        self.index = faiss.IndexFlatIP(m.shape[1]); self.index.add(m)
    def get(self, q, qv):
        k = ' '.join(q.strip().lower().split())
        if k in self.exact: self.stats['exact_hits'] += 1; e = self.exact[k]; return e['answer'], 'exact'
        if self.index is not None:
            qv = qv.astype('float32'); faiss.normalize_L2(qv)
            score, idx = self.index.search(qv, 1)
            if float(score[0, 0]) >= self.thr: self.stats['semantic_hits'] += 1; return self.entries[int(idx[0, 0])]['answer'], 'semantic'
        self.stats['misses'] += 1; return None, 'miss'
    def put(self, q, a, gen_time, qv):
        e = {'query': q, 'answer': a, 'gen_time': gen_time, 'emb': qv.reshape(-1).copy()}
        self.entries.append(e); self.exact[' '.join(q.strip().lower().split())] = e; self.rebuild()

# Fixed set: first 40 rows of the same 333-item evaluation set.
queries = eval_df['query'].tolist()[:40]
cache = SemanticCache(0.92)

def run_pass():
    rows = []
    for q in tqdm(queries, desc='Hybrid cache 40'):
        t = time.time(); qv = embed_texts([q]); hit, ht = cache.get(q, qv)
        if hit is not None:
            rows.append({'query': q, 'cache': ht, 'tier': 'cache', 'latency_sec': time.time() - t}); continue
        retrieved = hybrid_retrieve(q)
        samples = generate_slm(q, retrieved)
        rep, sc = rep_and_consistency(samples)
        # Use the learned SC-GRC router from the completed 333-item signal stage
        # (same router state chain used by Notebook 08).
        saved = sig[sig['query'] == q].iloc[0]
        feat = np.array([[1 - sc, 1 - float(saved['groundedness']), float(saved['cmi'])]], dtype='float32')
        escalate = bool(clf.predict(feat)[0]); prob = float(clf.predict_proba(feat)[0, 1])
        if escalate: ans = generate_llm(q, retrieved); tier = 'llm'
        else: ans = rep; tier = 'slm'
        gt = time.time() - t
        cache.put(q, ans, gt, qv)
        rows.append({'query': q, 'cache': 'miss', 'tier': tier, 'latency_sec': gt, 'self_consistency': sc, 'router_probability': prob})
    return rows

cold = run_pass(); warm = run_pass()
ca = np.mean([r['latency_sec'] for r in cold]); wa = np.mean([r['latency_sec'] for r in warm])
summary = {
    'N_queries': 40, 'threshold': 0.92,
    'cold_avg_latency_sec': float(ca), 'warm_avg_latency_sec': float(wa),
    'speedup_pct': float(100 * (ca - wa) / max(ca, 1e-9)),
    'exact_hits': int(cache.stats['exact_hits']), 'semantic_hits': int(cache.stats['semantic_hits']), 'misses': int(cache.stats['misses'])
}
json.dump(summary, open(os.path.join(WORK_STATE_DIR, 'sc_grc_cache40_summary.json'), 'w'), indent=2)
pd.DataFrame(cold).to_csv(os.path.join(WORK_STATE_DIR, 'sc_grc_cache40_cold.csv'), index=False)
pd.DataFrame(warm).to_csv(os.path.join(WORK_STATE_DIR, 'sc_grc_cache40_warm.csv'), index=False)
print('EXACTLY 40 QUERIES'); print(summary)
```
